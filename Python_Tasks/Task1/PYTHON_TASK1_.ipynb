```python
#35
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a) # this is shallow copy, as it creates a new outer list but the values of inner list are same so modifications on inner list affect both
```

```python
#36
import copy
a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a) # this is deepcopy, as it not only copies the outer list but also inner list so modifications on one list doesn't effect another list
```

```python
#37
import copy
a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b) #output: [[10], [20,30]]. As it shallow copy shares the outer list of both, so any changes to outer list will affect another
```

```python
#38
import copy
a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b) # output: [[10], [20]]. As deepcopy, creates the similar copy with new list for b, so chnages on a will not affect on b
```

```python
#39
import copy
a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b) # False. Because it only copies the outer list but not inner list which is not same, so its False
print(a[0] is b[0]) #True. Because it checks the values in their index positions
```
