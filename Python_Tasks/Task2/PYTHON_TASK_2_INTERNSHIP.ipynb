```python
#Q1. Find the Shortest Word in a Sentence
#Write a Python program to find the shortest word in a given sentence.
#Example input:     Python makes coding easy
# Expected output:      easy

sentence = "Python makes coding easy"
words = sentence.split()
shortest = words[0]

for word in words:
  if len(word) < len(shortest):
     shortest=word
print(shortest)
```

```python
#Q2. Find the Largest Number in a List
#Write a Python program to find the largest number in a list without using the built-in max() function.
#Example input:   [12, 45, 7, 89, 23]
#Expected output:    89

num = [12,45,7,89,23]
largest = num[0]

for i in num:
   if i > largest:
      largest = i

print(largest)
```

```python
#Q6. Count Vowels in a String
#Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.
#Example input:  Education
#Expected output: 5

text = "Education"
vowels = "aeiouAEIOU"
count = 0

for char in text:
    if char in vowels:
        count += 1

print(count)
```

```python
#Q7. Count Consonants in a String
#Write a Python program to count the consonants in a string. Ignore spaces, digits, and special characters.
#Example input:   Python 123!
#Expected output:   5

text = "Python 123!"
vowels = "aeiouAEIOU"
count = 0

for char in text:
    if char.isalpha() and char not in  vowels:
        count += 1

print(count)
```

```python
#Q8. Find the Most Frequent Character
#Write a Python program to find the character that appears most frequently in a string. If multiple
      #characters have the same highest frequency, return the first one encountered. Ignore spaces.
#Example input:   banana
#Expected output:  a


text = "banana"
text = text.replace(" ", "")
most_frequent = text[0]
for char in text:
    if text.count(char) > text.count(most_frequent):
        most_frequent = char

print(most_frequent)
```

```python
#Q9. Find the Least Frequent Character
#Write a Python program to find the character that appears least frequently in a string. If multiple characters have the same lowest frequency, return the first one encountered. Ignore spaces.
#Example input:success
#Expected output: u


text = "success"
text = text.replace(" ", "")
least_frequent = text[0]
for char in text:
    if text.count(char) < text.count(least_frequent):
        least_frequent = char

print(least_frequent)
```

```python
#Q10. Find All Characters Appearing More Than Once
#Write a Python program to identify all characters that appear more than once in a string. Display each
    # repeated character only once, in the order of its first appearance. Ignore spaces.
#Example input:  programming
#Expected output:   ['r', 'g', 'm']


text = "programming"
text = text.replace(" ", "")
repeated = []
for char in text:
    if text.count(char) > 1 and char not in repeated:
        repeated.append(char)

print(repeated)
```

```python
#Q11. Create a Dictionary of Words and Their First Characters
#Write a Python program to create a dictionary in which each word is a key and its first character is the value.
#Example input:Python Data Engineering
#Expected output: {'Python': 'P', 'Data': 'D', 'Engineering': 'E'}


text = "Python Data Engineering"
words = text.split()
result = {}
for word in words:
    result[word] = word[0]

print(result)
```

```python
#Q12. Create a Dictionary of Words and Their Last Characters
#Write a Python program to create a dictionary in which each word is a key and its last character is the value.
#Example input:Python Data Engineering
#Expected output:{'Python': 'n', 'Data': 'a', 'Engineering': 'g'}


text = "Python Data Engineering"
words = text.split()
result = {}
for word in words:
    result[word] = word[-1]

print(result)
```

```python
#Q13. Create a Dictionary of Words and Their Uppercase Versions
#Write a Python program to create a dictionary in which each word is a key and its uppercase version is the value.
#Example input:data science python
#Expected output: {'data': 'DATA', 'science': 'SCIENCE', 'python': 'PYTHON'}

names = "data science python"
words = names.split()
large = {}
for word in words:
    large[word] = word.upper()
print(large)
```

```python
#Q14. Create a Dictionary of Numbers and Their Squares
#Write a Python program to create a dictionary containing each number in a list as a key and its square as the value.
#Example input:[2, 3, 4, 5]
#Expected output:{2: 4, 3: 9, 4: 16, 5: 25}


numbers = [2, 3, 4, 5]
result = {}
for num in numbers:
    result[num] = num ** 2

print(result)
```

```python
#Q15. Create a Dictionary of Numbers and Their Even/Odd Status
#Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.
#Example input: [1, 2, 3, 4, 5]
#Expected output:  {1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}


numbers = [1, 2, 3, 4, 5]
result = {}
for num in numbers:
    if num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"
print(result)
```

```python
#Q16. Find the Most Frequent Word in a Sentence
#Write a Python program to find the word that appears most frequently in a sentence. Treat uppercase and lowercase versions of a word as the same word. If multiple words share the highest frequency, return the one that appears first.
#Example input:Python is easy and Python is powerful
#Expected output:python


text = "Python is easy and Python is powerful"
words = text.lower().split()
most_frequent = words[0]

for word in words:
    if words.count(word) > words.count(most_frequent):
        most_frequent = word

print(most_frequent)
```

```python
#Q17. Find All Words Appearing More Than Once
#Write a Python program to find all words that appear more than once in a sentence. Display each repeated word only once, in the order of its first appearance. Treat uppercase and lowercase versions as the same word.
#Example input:data science uses data and science tools
#Expected output:['data', 'science']


text = "data science uses data and science tools"
words = text.lower().split()
repeated = []
for word in words:
    if words.count(word) > 1 and word not in repeated:
        repeated.append(word)

print(repeated)
```

```python
#Q19. Count Words Whose Length Is Greater Than 4
#Write a Python program to count the words containing more than four characters in a sentence.
#Example input:Python makes problem solving interesting
#Expected output:4



text = "Python makes problem solving interesting"
words = text.split()
count = 0
for word in words:
    if len(word) > 4:
        count += 1

print(count)
```
