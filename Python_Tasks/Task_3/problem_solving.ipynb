# Q1. Word Frequency

## Pseudocode
1. Create an empty dictionary.
2. Split the sentence into words.
3. Visit each word one by one.
4. If the word already exists, increase its count by 1.
5. Otherwise, add the word with count 1.
6. Return the dictionary.

## Expected Output
{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}

```python

def word_frequency(sentence):
    frequency = {}

    for word in sentence.split():
        frequency[word] = frequency.get(word, 0) + 1

    return frequency


# Original test case
print(word_frequency("python is easy and python is powerful"))

# Extra test case
print(word_frequency("apple banana apple"))
```


# Q2. Character Frequency

## Pseudocode
1. Create an empty dictionary.
2. Visit each character in the word.
3. If the character already exists, increase its count by 1.
4. Otherwise, add the character with count 1.
5. Return the dictionary.

## Expected Output
{'b': 1, 'a': 3, 'n': 2}


```python

def character_frequency(word):
    frequency = {}

    for char in word:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    return frequency


# Original test case
print(character_frequency("banana"))

# Extra test case
print(character_frequency("hello"))
```


# Q3. First Non-Repeating Character

## Pseudocode
1. Count the frequency of each character.
2. Visit the characters in the original word from left to right.
3. Find the first character whose frequency is 1.
4. Return that character.
5. If no such character exists, return None.

## Expected Output
For "swiss", the output is 'w'.

```python

def first_non_repeating(word):
    frequency = {}

    for char in word:
        frequency[char] = frequency.get(char, 0) + 1

    for char in word:
        if frequency[char] == 1:
            return char

    return None


# Original test case
print(first_non_repeating("swiss"))

# Extra test case
print(first_non_repeating("aabb"))
```


# Q4. Check Anagrams

## Pseudocode
1. Take two words as input.
2. Sort the characters of both words.
3. Compare the sorted results.
4. If both are equal, return True.
5. Otherwise, return False.

## Expected Output
"listen" and "silent" → True

```python

def are_anagrams(word1, word2):
    return sorted(word1) == sorted(word2)


# Original test case
print(are_anagrams("listen", "silent"))

# Extra test case
print(are_anagrams("hello", "world"))
```


# Q5. First Repeating Character

## Pseudocode
1. Create an empty set to store visited characters.
2. Visit each character from left to right.
3. If the character is already in the set, return it.
4. Otherwise, add the character to the set.
5. If no character repeats, return None.

## Expected Output
"programming" → 'r'

```python

def first_repeating(word):
    seen = set()

    for char in word:
        if char in seen:
            return char
        seen.add(char)

    return None


# Original test case
print(first_repeating("programming"))

# Extra test case
print(first_repeating("abcde"))
```


# Q6. Move Zeros to the End

## Pseudocode
1. Create a list containing all non-zero numbers.
2. Create another list containing all zeros.
3. Combine both lists.
4. Return the combined list.

## Expected Output
[0, 3, 0, 1, 2, 0, 4] → [3, 1, 2, 4, 0, 0, 0]

```python

def move_zeros(numbers):
    non_zeros = [num for num in numbers if num != 0]
    zeros = [num for num in numbers if num == 0]

    return non_zeros + zeros


# Original test case
print(move_zeros([0, 3, 0, 1, 2, 0, 4]))

# Extra test case
print(move_zeros([0, 1, 0, 2]))
```


# Q7. Second-Highest Distinct Number

## Pseudocode
1. Remove duplicate numbers from the list.
2. Sort the unique numbers in descending order.
3. If fewer than two unique numbers exist, return None.
4. Otherwise, return the number at index 1.

## Expected Output
[10, 20, 40, 30, 40, 20] → 30

```python

def second_highest(numbers):
    unique_numbers = sorted(set(numbers), reverse=True)

    if len(unique_numbers) < 2:
        return None

    return unique_numbers[1]


# Original test case
print(second_highest([10, 20, 40, 30, 40, 20]))

# Extra test case
print(second_highest([5, 5, 3, 3, 1]))
```


# Q8. Remove Duplicates While Preserving Order

## Pseudocode
1. Create an empty result list.
2. Visit each number in the original list.
3. If the number is not already in the result list, add it.
4. Return the result list.

## Expected Output
[4, 2, 4, 5, 2, 7, 5] → [4, 2, 5, 7]

```python

def remove_duplicates(numbers):
    result = []

    for num in numbers:
        if num not in result:
            result.append(num)

    return result


# Original test case
print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))

# Extra test case
print(remove_duplicates([1, 1, 2, 3, 2]))
```


# Q9. Find the Missing Number

## Pseudocode
1. Take a list containing numbers from 1 to N, with one number missing.
2. Calculate the expected sum from 1 to N using N * (N + 1) / 2.
3. Calculate the sum of the given list.
4. Subtract the actual sum from the expected sum.
5. Return the missing number.

## Expected Output
List: [1, 2, 4, 5, 6]
N: 6
Missing number: 3

```python

def missing_number(numbers, n):
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(numbers)

    return expected_sum - actual_sum


# Original test case
print(missing_number([1, 2, 4, 5, 6], 6))

# Extra test case
print(missing_number([1, 2, 3, 5], 5))
```


# Q10. Two Sum

## Pseudocode
1. Take a list of numbers and a target value.
2. Select each number as the second number of a pair.
3. Check the numbers before it one by one.
4. If the sum of the two numbers equals the target, return the pair.
5. If no matching pair exists, return an empty list.

## Expected Output
Numbers: [2, 7, 11, 15]
Target: 9
Result: [2, 7]

```python

def two_sum(numbers, target):
    for j in range(len(numbers)):
        for i in range(j):
            if numbers[i] + numbers[j] == target:
                return [numbers[i], numbers[j]]

    return []


# Original test case
print(two_sum([2, 7, 11, 15], 9))

# Extra test case
print(two_sum([3, 2, 4], 6))
```


# Q11. Number Closest to Zero

## Pseudocode
1. Check whether the list is empty.
2. If it is empty, return None.
3. Otherwise, find the number with the smallest absolute value.
4. If two numbers are equally close to zero, choose the positive number.
5. Return the selected number.

## Expected Output
[-8, -3, 2, 5, -2] → 2

```python

def closest_to_zero(numbers):
    if not numbers:
        return None

    return min(numbers, key=lambda num: (abs(num), num < 0))


# Original test case
print(closest_to_zero([-8, -3, 2, 5, -2]))

# Extra test case
print(closest_to_zero([-4, 4, 7]))
```


# Q12. Common Elements in Two Lists

## Pseudocode
1. Create an empty result list.
2. Visit each number in the first list.
3. Check whether the number exists in the second list.
4. Check whether it is already in the result list.
5. If both conditions are satisfied, add it to the result list.
6. Return the result list.

## Expected Output
List 1: [1, 2, 3, 4, 5]
List 2: [3, 4, 5, 6, 7]
Result: [3, 4, 5]

```python

def common_elements(list1, list2):
    result = []

    for num in list1:
        if num in list2 and num not in result:
            result.append(num)

    return result


# Original test case
print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))

# Extra test case
print(common_elements([1, 2, 2, 3], [2, 3, 4]))
```


# Q13. Rotate a List Right by K Positions

## Pseudocode
1. Check whether the list is empty.
2. If it is empty, return an empty list.
3. Calculate K modulo the length of the list.
4. Take the last K elements.
5. Take the remaining elements.
6. Join the last K elements before the remaining elements.
7. Return the rotated list.

## Expected Output
List: [1, 2, 3, 4, 5]
K: 2
Result: [4, 5, 1, 2, 3]

```python

def rotate_right(numbers, k):
    if not numbers:
        return []

    k = k % len(numbers)

    if k == 0:
        return numbers[:]

    return numbers[-k:] + numbers[:-k]


# Original test case
print(rotate_right([1, 2, 3, 4, 5], 2))

# Extra test case
print(rotate_right([1, 2, 3, 4], 1))
```


# Q14. Maximum Sum of Adjacent Elements

## Pseudocode
1. Check whether the list contains at least two numbers.
2. If not, return None.
3. Take each pair of neighboring numbers.
4. Calculate the sum of each pair.
5. Find the maximum sum.
6. Return the maximum sum.

## Expected Output
List: [4, 7, 2, 9, 5]
Result: 14

```python

def max_adjacent_sum(numbers):
    if len(numbers) < 2:
        return None

    return max(
        numbers[i] + numbers[i + 1]
        for i in range(len(numbers) - 1)
    )


# Original test case
print(max_adjacent_sum([4, 7, 2, 9, 5]))

# Extra test case
print(max_adjacent_sum([1, 5, 3, 8]))
```


# Q15. Find All Unique Pairs with Target Sum

## Pseudocode
1. Create an empty set to store unique pairs.
2. Select each number from the list.
3. Compare it with the numbers that come after it.
4. If the sum equals the target, sort the two numbers and store the pair.
5. Use a set to avoid duplicate pairs.
6. Sort all pairs and convert them into lists.
7. Return the result.

## Expected Output
Numbers: [2, 4, 3, 5, 7, 8, 1]
Target: 9
Result: [[1, 8], [2, 7], [4, 5]]

```python

def all_target_pairs(numbers, target):
    pairs = set()

    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                pair = tuple(sorted((numbers[i], numbers[j])))
                pairs.add(pair)

    return [list(pair) for pair in sorted(pairs)]


# Original test case
print(all_target_pairs([2, 4, 3, 5, 7, 8, 1], 9))

# Extra test case
print(all_target_pairs([1, 2, 3, 4, 5], 6))
```
